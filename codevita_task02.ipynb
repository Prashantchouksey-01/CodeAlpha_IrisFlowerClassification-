```python
print(df[['Unemployment_Percentage',
          'Unemployment_Rate']].corr())
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Handle missing values
df = df.dropna(subset=['State', 'Unemployment_Percentage'])

# Convert percentage column to numeric
df['Unemployment_Percentage'] = pd.to_numeric(
    df['Unemployment_Percentage'], errors='coerce'
)

df = df.dropna(subset=['Unemployment_Percentage'])

print("Cleaned Dataset Shape:", df.shape)
```

```python
highest = df.loc[df['Unemployment_Percentage'].idxmax()]

print("Highest Unemployment State:", highest['State'])
print("Unemployment Rate:", highest['Unemployment_Percentage'], "%")
```

```python
lowest = df.loc[df['Unemployment_Percentage'].idxmin()]

print("Lowest Unemployment State:", lowest['State'])
print("Unemployment Rate:", lowest['Unemployment_Percentage'], "%")
```

```python
sorted_df = df.sort_values(
    by='Unemployment_Percentage',
    ascending=False
)

display(sorted_df[['State', 'Unemployment_Percentage']])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 10))

data = df.sort_values('Unemployment_Percentage')

plt.barh(
    data['State'],
    data['Unemployment_Percentage'],
    color='teal'
)

plt.xlabel('Unemployment Rate (%)')
plt.ylabel('State')
plt.title('Unemployment Rate by State - December 2022')
plt.grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
top10 = df.nlargest(10, 'Unemployment_Percentage')

plt.figure(figsize=(10, 6))

plt.bar(
    top10['State'],
    top10['Unemployment_Percentage'],
    color='tomato'
)

plt.xticks(rotation=45, ha='right')
plt.xlabel('State')
plt.ylabel('Unemployment (%)')
plt.title('Top 10 States with Highest Unemployment')

plt.tight_layout()
plt.show()
```

```python
top5 = df.nlargest(5, 'Unemployment_Percentage')

plt.figure(figsize=(8, 8))

plt.pie(
    top5['Unemployment_Percentage'],
    labels=top5['State'],
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Top 5 States - Unemployment Rate Comparison')
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.hist(
    df['Unemployment_Percentage'],
    bins=10,
    color='orange',
    edgecolor='black'
)

plt.xlabel('Unemployment Rate (%)')
plt.ylabel('Number of States')
plt.title('Distribution of Unemployment Rates')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(df['Unemployment_Percentage'])

plt.ylabel('Unemployment Rate (%)')
plt.title('Unemployment Rate - Outlier Detection')

plt.show()
```

```python
print("========== FINAL ANALYSIS ==========")

print("Total States:", df['State'].nunique())

print("Average Unemployment:",
      round(df['Unemployment_Percentage'].mean(), 2), "%")

print("Maximum Unemployment:",
      df['Unemployment_Percentage'].max(), "%")

print("Minimum Unemployment:",
      df['Unemployment_Percentage'].min(), "%")

print("Median Unemployment:",
      df['Unemployment_Percentage'].median(), "%")

print("Standard Deviation:",
      round(df['Unemployment_Percentage'].std(), 2))
```

```python
highest = df.loc[df['Unemployment_Percentage'].idxmax()]
lowest = df.loc[df['Unemployment_Percentage'].idxmin()]

average = df['Unemployment_Percentage'].mean()

print(f"""
CONCLUSION:

1. The average unemployment rate is {average:.2f}%.
2. {highest['State']} has the highest unemployment rate
   of {highest['Unemployment_Percentage']}%.
3. {lowest['State']} has the lowest unemployment rate
   of {lowest['Unemployment_Percentage']}%.
4. Unemployment rates vary across different states.
5. The analysis highlights regional differences
   in unemployment during December 2022.
""")
```
